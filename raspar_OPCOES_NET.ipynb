```python
# =============================
# Coletar Opcoes Net
# =============================
def opcoes_net():
    import time
    from bs4 import BeautifulSoup
    import undetected_chromedriver as uc
    from selenium.webdriver.common.by import By
    from selenium.webdriver.support import expected_conditions as EC
    from selenium.webdriver.support.ui import WebDriverWait
    
    # ============================================================
    # CONFIGURAÇÕES
    # ============================================================
    CPF_USUARIO = "350.422.188-74"
    SENHA_USUARIO = "logoi777"
    acao = "PETR4"
    
    # ============================================================
    # CONFIGURA CHROME
    # ============================================================
    options = uc.ChromeOptions()
    
    # Mantém browser ativo, mas fora da tela
    options.add_argument("--window-position=-2000,0") # remover esse para visualizar browser
    options.add_argument("--window-size=1920,1080")
    
    # Evita congelamento em segundo plano
    options.add_argument("--disable-background-timer-throttling")
    options.add_argument("--disable-backgrounding-occluded-windows")
    options.add_argument("--disable-renderer-backgrounding")
    
    driver = uc.Chrome(options=options, version_main=150)
    
    driver.set_page_load_timeout(60)
    wait = WebDriverWait(driver, 20)
    
    # ============================================================
    # LOGIN
    # ============================================================
    try:
        print("1. Acessando página de ações...")
        driver.get("https://opcoes.net.br/acoes")
    
        # --------------------------------------------------------
        # CLICAR LOGIN
        # --------------------------------------------------------
        while True:
            try:
                botao = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'a.btn.btn-primary[href*="/login"]')))
                driver.execute_script("arguments[0].click();", botao)
    
                wait.until(lambda d: "/login" in d.current_url)
                break
    
            except Exception as e:
                print("Esperando botão login...", e)
                time.sleep(2)
    
        # --------------------------------------------------------
        # CPF
        # --------------------------------------------------------
        print("2. Informando CPF...")
    
        campo_cpf = wait.until(EC.visibility_of_element_located((By.ID, "CPF")))
        campo_cpf.clear()
        campo_cpf.send_keys(CPF_USUARIO)
    
        # --------------------------------------------------------
        # SENHA
        # --------------------------------------------------------
        print("3. Informando senha...")
    
        campo_senha = wait.until(EC.visibility_of_element_located((By.ID, "Password")))
        campo_senha.clear()
        campo_senha.send_keys(SENHA_USUARIO)
    
        # --------------------------------------------------------
        # BOTÃO ENTRAR
        # --------------------------------------------------------
        print("4. Efetuando login...")
    
        botao_entrar = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "button.btn.btn-default[type='submit']")))
        driver.execute_script("arguments[0].click();", botao_entrar)
        time.sleep(6)
    
        # --------------------------------------------------------
        # IR PARA AÇÕES
        # --------------------------------------------------------
        if "/acoes" not in driver.current_url:
    
            print("Redirecionando para ações...")
            driver.get("https://opcoes.net.br/acoes")
    
        wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "table tbody tr")))
        time.sleep(3)
    
        # --------------------------------------------------------
        # FECHAR POPUP
        # --------------------------------------------------------
        try:
    
            botao_modal = WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//a[contains(text(),'Continuar com dados de fechamento')]")))
            driver.execute_script( "arguments[0].click();", botao_modal)
    
        except Exception:
            pass
    
    except Exception as e:
    
        print("Erro no login:", e)
    
    
    # ===========================================================
    # ABRIR ATIVO
    # ===========================================================
    print(f"Buscando ativo {acao}")
    
    while True:
        try:
            xpath_acao = (f"//table//tbody//td"
                f"[normalize-space(text())='{acao}']")
    
    
            elemento_acao = wait.until(
                EC.element_to_be_clickable(
                    (By.XPATH, xpath_acao)))
    
            driver.execute_script(
                "arguments[0].click();", elemento_acao)
    
            print(            "Ativo aberto com sucesso:", acao)
            break
    
        except Exception as e:
            print(            "Aguardando ativo...", e)
            time.sleep(2)
    
    print("Página carregada com sucesso")

    # ======================================
    # COlETAR TABELA 
    # ======================================

    from selenium.webdriver.common.by import By
    
    # Captura o texto do elemento localizado pelo XPath
    elemento = driver.find_element(By.XPATH, "/html/body/div[1]/div[1]/div[1]")
    dados = elemento.text
    
    #print(dados)
    
    import pandas as pd
    import re
    
    # Definição das colunas
    colunas = [
        "Ativo", "Var. %", "Últ.", "Data",
        "CALLs_IV_Rank", "CALLs_Perc", "CALLs_Vol_Impl",
        "Diff_Vol",
        "PUTs_IV_Rank", "PUTs_Perc", "PUTs_Vol_Impl",
        "HV_Rank", "Perc_Hist", "Vol_Hist",
        "Volume_Financeiro"
    ]
    
    linhas_processadas = []
    
    # Expressão regular ajustada para pegar apenas tickers válidos da B3 (ex: YDUQ3, WEGE3, TAEE11)
    patrao_ativo = re.compile(r'^[A-Z]{4}\d{1,2}\b')
    
    for linha in dados.split('\n'):
        linha = linha.strip()
        partes = linha.split()
        
        # Valida se a linha começa com um ticker E se possui no mínimo 5 elementos
        if partes and patrao_ativo.match(partes[0]) and len(partes) >= 5:
            if len(partes) == 15:
                linhas_processadas.append(partes)
            else:
                # Captura segura dos elementos fixos
                ativo, var, ult, data = partes[0], partes[1], partes[2], partes[3]
                volume = partes[-1]
                meio = partes[4:-1]
                
                # Preenche o meio com None para manter o alinhamento correto em 15 colunas
                meio_com_nulos = meio + [None] * (10 - len(meio))
                linhas_processadas.append([ativo, var, ult, data] + meio_com_nulos + [volume])
    
    # Criando o DataFrame
    df_acoes = pd.DataFrame(linhas_processadas, columns=colunas)
    
    df_ativos_b3_ticker = df_acoes['Ativo'].to_list()
    #display(df_ativos_b3_ticker)
    display(df_acoes)

    # ==============================================================
    # MUDAR PAGINA PARA #https://opcoes.net.br/opcoes/bovespa/{acao}
    driver.get(f"https://opcoes.net.br/opcoes/bovespa/{acao}")

    # CRIAR AQUI LOOPING FOR PARA COLETAR TODAS AS CÇÕES >>>>--------------->
    # ==============================================================
    
    # CLICAR EM TODOS OS VENCIMENTOS DE OPÇOES ===========================
    time.sleep(10)
    from selenium.webdriver.common.by import By
    import time
    
    checkboxes = driver.find_elements(
        By.XPATH,
        "//input[@type='checkbox']"
    )
    
    print(f"{len(checkboxes)} checkboxes encontrados.")
    
    for i, cb in enumerate(checkboxes, start=1):
        try:
            driver.execute_script("arguments[0].scrollIntoView({block:'center'});", cb)
            time.sleep(0.1)
    
            if not cb.is_selected():
                driver.execute_script("arguments[0].click();", cb)
    
            print(f"{i} - {cb.get_attribute('value')} marcado.")
    
        except Exception as e:
            print(f"Erro no checkbox {i}: {e}")
    
    #####################################################################
    # Abrir Todo o Range de Strikes
    # Executa JavaScript no slider
    script = """
    var slider = $("#strike-range");
    
    var min = slider.slider("option", "min");
    var max = slider.slider("option", "max");
    
    slider.slider("values", [min, max]);
    
    slider.slider("option", "slide").call(
        slider,
        null,
        {
            values: [min, max]
        }
    );
    
    slider.trigger("slidechange");
    """
    
    driver.execute_script(script)
    time.sleep(10)

    # ================================
    # COLETAR TABELA =================
    # ================================

    from bs4 import BeautifulSoup
    import pandas as pd
    
    # UMA chamada só pro Selenium
    html = driver.find_element(By.ID, "tblListaOpc").get_attribute("outerHTML")
    
    soup = BeautifulSoup(html, "lxml")  # ou "html.parser" se não tiver lxml
    
    # Headers (primeira linha do thead)
    headers = [th.get_text(strip=True) for th in soup.select("thead tr:first-child th")]
    
    # Linhas do tbody
    dados = []
    for tr in soup.select("tbody tr"):
        celulas = [td.get_text(strip=True) for td in tr.find_all("td")]
        if celulas:  # ignora linhas vazias
            dados.append(celulas)
    
    df_tabela = pd.DataFrame(dados, columns=headers)

    # reduzir tabela
    df_tabela.loc[:, :'Vega']
    # Remover Colunas
    #df_tabela = df_tabela.drop(columns=["F.M."])

    # Exibir Primeira coleta
    #display(df_tabela)
    # =======================================================
    # FORMATAR TABELA
    # =======================================================

    colunas_final = ['Ticker', 'F.M.','Tipo', 'Dias úteis','Mod.', 'Strike', 'A/I/OTM', 'Dist. (%) do Strike', 'Último', 'Var.\xa0(%)', 'Núm. de Neg.',
                     'Vol. Financeiro', 'Vol. Impl. (%)', 'Delta', 'Gamma', 'Theta ($)', 'Theta (%)', 'Vega']

    colunas_converter = ['Strike',  'Dist. (%) do Strike', 'Último', 'Var.\xa0(%)', 'Núm. de Neg.',
                     'Vol. Financeiro', 'Vol. Impl. (%)', 'Delta', 'Gamma', 'Theta ($)', 'Theta (%)', 'Vega']

    # Converter tipos
    df_tabela[colunas_converter] = (
        df_tabela[colunas_converter]
        .apply(
            lambda col: (
                col.astype(str)
                .str.replace(r'\.(?=\d{3}(?:,|$))', '', regex=True)
                .str.replace(',', '.', regex=False)
                .str.replace('%', '', regex=False)
                .str.strip()
                .replace({'': None, '-': None, 'nan': None})
            )
        )
        .apply(pd.to_numeric, errors='coerce')
    )
    
    # Definir Tabela
    df_tabela = df_tabela[colunas_final]
    
    # Exibir as colunas desejadas no final
    display(df_tabela)
    
    # Conferir tipos somente delas
    df_tabela.dtypes
    driver.quit()


    # Filtrar Tabel Opções
    df_tabela = df_tabela[df_tabela.loc[:, 'Último':].notna().all(axis=1)]

    with pd.ExcelWriter("df_cotacoes.xlsx", engine="openpyxl", mode="a", if_sheet_exists="replace") as writer:
        df_acoes.to_excel(writer, sheet_name="tabela_opcoes", index=False)
        
    return df_tabela, df_acoes
```

```python
import os
import time
import uuid
import shutil
import tempfile
import pandas as pd
import undetected_chromedriver as uc

from bs4 import BeautifulSoup
from concurrent.futures import ThreadPoolExecutor, as_completed
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait


ARQUIVO_EXCEL = "df_cotacoes.xlsx"
MAX_WORKERS = max(1, (os.cpu_count() or 1) // 2)
TENTATIVAS = 3
TIMEOUT_PAGINA = 60
TIMEOUT_ELEMENTO = 30
TEMPO_ENTRE_TENTATIVAS = 5
LOCK_FILE = ARQUIVO_EXCEL + ".lock"


class ExcelFileLock:
    def __init__(self, lock_file, timeout=300, stale_time=1800):
        self.lock_file = lock_file
        self.timeout = timeout
        self.stale_time = stale_time
        self.lock_acquired = False

    def acquire(self):
        inicio = time.time()

        while True:
            try:
                fd = os.open(
                    self.lock_file,
                    os.O_CREAT | os.O_EXCL | os.O_WRONLY
                )

                with os.fdopen(fd, "w") as f:
                    f.write(f"PID={os.getpid()}\nTIME={time.time()}")

                self.lock_acquired = True
                return

            except FileExistsError:
                try:
                    if time.time() - os.path.getmtime(self.lock_file) > self.stale_time:
                        os.remove(self.lock_file)
                        continue
                except FileNotFoundError:
                    continue

                if time.time() - inicio > self.timeout:
                    raise TimeoutError(f"Timeout aguardando {self.lock_file}")

                time.sleep(2)

    def release(self):
        if self.lock_acquired:
            try:
                os.remove(self.lock_file)
            except FileNotFoundError:
                pass
            self.lock_acquired = False

    def __enter__(self):
        self.acquire()
        return self

    def __exit__(self, *args):
        self.release()


def criar_driver():
    options = uc.ChromeOptions()
    options.add_argument("--window-position=-2000,0")
    options.add_argument("--window-size=1920,1080")
    options.add_argument("--disable-background-timer-throttling")
    options.add_argument("--disable-backgrounding-occluded-windows")
    options.add_argument("--disable-renderer-backgrounding")
    options.add_argument("--disable-popup-blocking")
    options.add_argument("--disable-notifications")

    user_data_dir = tempfile.mkdtemp(
        prefix=f"chrome_opcoes_{uuid.uuid4().hex}_"
    )
    options.add_argument(f"--user-data-dir={user_data_dir}")

    try:
        driver = uc.Chrome(options=options)
        driver.set_page_load_timeout(TIMEOUT_PAGINA)
        return driver, user_data_dir
    except Exception:
        shutil.rmtree(user_data_dir, ignore_errors=True)
        raise


def formatar_tabela(df):
    colunas_final = [
        'Ticker', 'F.M.', 'Tipo', 'Dias úteis', 'Mod.', 'Strike',
        'A/I/OTM', 'Dist. (%) do Strike', 'Último', 'Var.\xa0(%)',
        'Núm. de Neg.', 'Vol. Financeiro', 'Vol. Impl. (%)',
        'Delta', 'Gamma', 'Theta ($)', 'Theta (%)', 'Vega'
    ]

    colunas_converter = [
        'Strike', 'Dist. (%) do Strike', 'Último', 'Var.\xa0(%)',
        'Núm. de Neg.', 'Vol. Financeiro', 'Vol. Impl. (%)',
        'Delta', 'Gamma', 'Theta ($)', 'Theta (%)', 'Vega'
    ]

    faltantes = [c for c in colunas_final if c not in df.columns]

    if faltantes:
        raise ValueError(f"Colunas faltantes: {faltantes}")

    for coluna in colunas_converter:
        df[coluna] = (
            df[coluna]
            .astype(str)
            .str.replace(r'\.(?=\d{3}(?:,|$))', '', regex=True)
            .str.replace(',', '.', regex=False)
            .str.replace('%', '', regex=False)
            .str.strip()
            .replace({'': None, '-': None, 'nan': None})
        )

        df[coluna] = pd.to_numeric(
            df[coluna],
            errors="coerce"
        )

    df = df[colunas_final].copy()

    df = df[
        df.loc[:, 'Último':].notna().all(axis=1)
    ].copy()

    return df


def coletar_ticker(ticker):
    ultimo_erro = None

    for tentativa in range(1, TENTATIVAS + 1):
        driver = None
        user_data_dir = None

        try:
            print(f"[{ticker}] tentativa {tentativa}/{TENTATIVAS}")

            driver, user_data_dir = criar_driver()
            wait = WebDriverWait(driver, TIMEOUT_ELEMENTO)

            driver.get(
                f"https://opcoes.net.br/opcoes/bovespa/{ticker}"
            )

            try:
                wait.until(
                    EC.presence_of_element_located(
                        (By.XPATH, "//input[@type='checkbox']")
                    )
                )
            except Exception:
                time.sleep(5)

            checkboxes = driver.find_elements(
                By.XPATH,
                "//input[@type='checkbox']"
            )

            if not checkboxes:
                raise RuntimeError("Nenhum checkbox encontrado")

            for cb in checkboxes:
                try:
                    driver.execute_script(
                        "arguments[0].scrollIntoView({block:'center'});",
                        cb
                    )

                    if not cb.is_selected():
                        driver.execute_script(
                            "arguments[0].click();",
                            cb
                        )

                except Exception:
                    pass

            driver.execute_script("""
                var slider = $("#strike-range");

                if (!slider.length)
                    throw new Error("Slider não encontrado");

                var min = slider.slider("option", "min");
                var max = slider.slider("option", "max");

                slider.slider("values", [min, max]);

                slider.slider("option", "slide").call(
                    slider,
                    null,
                    {values: [min, max]}
                );

                slider.trigger("slidechange");
            """)

            wait.until(
                EC.presence_of_element_located(
                    (By.ID, "tblListaOpc")
                )
            )

            time.sleep(3)

            html = driver.find_element(
                By.ID,
                "tblListaOpc"
            ).get_attribute("outerHTML")

            if not html:
                raise RuntimeError("HTML da tabela vazio")

            soup = BeautifulSoup(html, "lxml")

            headers = [
                th.get_text(strip=True)
                for th in soup.select(
                    "thead tr:first-child th"
                )
            ]

            dados = [
                [
                    td.get_text(strip=True)
                    for td in tr.find_all("td")
                ]
                for tr in soup.select("tbody tr")
            ]

            dados = [linha for linha in dados if linha]

            if not headers or not dados:
                raise RuntimeError("Tabela sem dados")

            df = pd.DataFrame(
                dados,
                columns=headers
            )

            df = formatar_tabela(df)

            print(f"OK [{ticker}] {len(df)} linhas")

            return ticker, df, None

        except Exception as e:
            ultimo_erro = e
            print(f"ERRO [{ticker}] tentativa {tentativa}: {e}")

            if tentativa < TENTATIVAS:
                time.sleep(TEMPO_ENTRE_TENTATIVAS)

        finally:
            if driver is not None:
                try:
                    driver.quit()
                except Exception:
                    pass

            if user_data_dir:
                shutil.rmtree(
                    user_data_dir,
                    ignore_errors=True
                )

    return ticker, None, repr(ultimo_erro)


def salvar_excel(df, ticker):
    with ExcelFileLock(LOCK_FILE):
        if os.path.exists(ARQUIVO_EXCEL):
            with pd.ExcelWriter(
                ARQUIVO_EXCEL,
                engine="openpyxl",
                mode="a",
                if_sheet_exists="replace"
            ) as writer:
                df.to_excel(
                    writer,
                    sheet_name=f"opcoes_{ticker}",
                    index=False
                )
        else:
            with pd.ExcelWriter(
                ARQUIVO_EXCEL,
                engine="openpyxl",
                mode="w"
            ) as writer:
                df.to_excel(
                    writer,
                    sheet_name=f"opcoes_{ticker}",
                    index=False
                )


tickers = df_cotacoes.columns.to_list()

inicio = time.time()
resultados = {}
erros = {}

print(f"{len(tickers)} tickers | {MAX_WORKERS} workers")

with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:

    futures = {
        executor.submit(coletar_ticker, ticker): ticker
        for ticker in tickers
    }

    for future in as_completed(futures):
        ticker = futures[future]

        try:
            ticker, df, erro = future.result()

            if df is not None:
                salvar_excel(df, ticker)
                resultados[ticker] = df
                print(f"SALVO [{ticker}]")
            else:
                erros[ticker] = erro
                print(f"FALHOU [{ticker}]")

        except Exception as e:
            erros[ticker] = repr(e)
            print(f"ERRO [{ticker}]: {e}")


print(f"Concluído em {time.time() - inicio:.1f}s")
```

```python
def cotacoes_mt5_loop(tickers, timeframe=mt5.TIMEFRAME_D1, n_barras=1000):

    import pandas as pd
    import MetaTrader5 as mt5
    import time

    if not mt5.initialize():
        raise RuntimeError(
            f"Erro ao inicializar MetaTrader5: {mt5.last_error()}"
        )

    resultados = []

    for ticker in tickers:

        print(f"Coletando: {ticker}")

        try:

            # Selecionar ativo
            if not mt5.symbol_select(ticker, True):
                print(f"Ticker não encontrado: {ticker}")
                continue

            # Pequena pausa para o terminal processar
            time.sleep(0.2)

            # Buscar candles
            rates = mt5.copy_rates_from_pos(
                ticker,
                timeframe,
                0,
                n_barras
            )

            if rates is None or len(rates) == 0:
                print(
                    f"Sem dados para {ticker}: "
                    f"{mt5.last_error()}"
                )
                continue

            df = pd.DataFrame(rates)

            # Converter tempo
            df["time"] = pd.to_datetime(
                df["time"],
                unit="s"
            )

            df["ticker"] = ticker

            resultados.append(df)

        except Exception as e:

            print(
                f"Erro em {ticker}: "
                f"{type(e).__name__}: {e}"
            )

    mt5.shutdown()

    if not resultados:
        return pd.DataFrame()

    return pd.concat(
        resultados,
        ignore_index=True
    )
```

```python
import MetaTrader5 as mt5
def cotacoes_mt5_loop(tickers, timeframe=mt5.TIMEFRAME_D1, n_barras=1000):

    # Apagar Diretorio # ============================================>>>>>>>>>>>>>>>>
    '''from pathlib import Path 
    import shutil
    pasta = Path(r"C:\Users\Usuário\AppData\Roaming\MetaQuotes\Terminal\D0E8209F77C8CF37AD8BF550E51FF075\bases\XPMT5-DEMO\history")
    
    for item in pasta.iterdir():
        if item.is_dir():
            shutil.rmtree(item)
        else:
            item.unlink()
    
    print("Conteúdo removido com sucesso.") # ============================================>>>>>>>>>>>>>>>>'''

    
    import pandas as pd
    from IPython.display import clear_output
    import time
    
    df_lista = pd.DataFrame()
    for i in tickers:
        #print(i)
        
        if not mt5.initialize():
            raise RuntimeError(
                f"Erro ao conectar no MT5:\n{mt5.last_error()}")
    
        '''if not mt5.symbol_select(i, True):
            mt5.shutdown()
            raise ValueError(f"Ticker '{i}' não encontrado.")'''
            
    
        rates = mt5.copy_rates_from_pos(i, timeframe, 0, n_barras)
    
        #mt5.shutdown()
    
        if rates is None:
            raise ValueError("Nenhum dado retornado.")
    
        df = pd.DataFrame(rates)
    
        df["time"] = pd.to_datetime(
            df["time"],
            unit="s"
        )
    
        df.rename(
            columns={
                "time":"Data",
                "open":"Open",
                "high":"High",
                "low":"Low",
                "close":i,
                "tick_volume":"Volume",
                "real_volume":"RealVolume",
                "spread":"Spread"}, inplace=True)
    
        df.set_index("Data", inplace=True)
        df = df.iloc[:,3]
        #print(df)
        df.bfill()
        df_lista = pd.concat([df_lista, df], axis=1)

    # Fechar MT5
    import subprocess
    import time
    
    mt5.shutdown()
    import gc
    import time
    gc.collect()
    time.sleep(10)

    r = subprocess.run(
        ["taskkill", "/F", "/IM", "terminal64.exe"],
        capture_output=True,
        text=True
    )

    print(r.stdout)
    print(r.stderr)
    time.sleep(5)

    
    # Apagar Diretorio # ============================================>>>>>>>>>>>>>>>>
    from pathlib import Path 
    import shutil
    pasta = Path(r"C:\Users\Usuário\AppData\Roaming\MetaQuotes\Terminal\D0E8209F77C8CF37AD8BF550E51FF075\bases\XPMT5-DEMO\history")
    
    for item in pasta.iterdir():
        if item.is_dir():
            shutil.rmtree(item)
        else:
            item.unlink()
    
    print("Conteúdo removido com sucesso.") # ============================================>>>>>>>>>>>>>>>>
    
    return df_lista
```

```python
import yfinance as yf
import pandas as pd

def cotacoes_yf(codigos, tf='1d', cotacoes=756, NaN_Maximo=0.2, PrimeiraData='2020-01-01'):
    df_final = pd.DataFrame()
    ticks_validos = []
    ticks_invalidos = []

    for ticker in codigos:
        try:
            df = yf.download(ticker, period=f'{cotacoes}d', interval=tf, progress=False)
            if df.empty or df.isna().mean().mean() > NaN_Maximo:
                ticks_invalidos.append(ticker)
                continue
            df = df[['Close']].rename(columns={'Close': ticker[:-3]})
            df.index.name = 'time'
            df_final = pd.concat([df_final, df], axis=1)
            ticks_validos.append(ticker)
        except Exception:
            ticks_invalidos.append(ticker)

    df_final = df_final[df_final.index >= PrimeiraData]

    return df_final, ticks_validos, ticks_invalidos

cotacoes_yf(tickers, tf='1d', cotacoes=756, NaN_Maximo=0.2, PrimeiraData='2024-01-01')
```

```python
def loog_classificacao_vol(tickers):
    import pandas as pd
    import numpy as np
    
    # Coletar cotacoes
    df_lista_cotacoes = cotacoes_mt5_lista_2(tickers, timeframe=mt5.TIMEFRAME_D1, n_barras=600)
    df_lista_cotacoes.columns.to_list()
    
    df_class_vol = pd.DataFrame()
    for i in df_lista_cotacoes:
        #df = cotacoes_mt5_OHLC(i, timeframe=mt5.TIMEFRAME_D1, n_barras=600)
        df = pd.DataFrame(df_lista_cotacoes[i])

        #df = cotacoes_yfinance(i, start_data)
        df['ticker'] = i
    
        # ============================================================
        # CONFIGURAÇÕES
        # ============================================================
    
        JANELA_VOL = 21
        JANELA_IV = 252
        DIAS_ANO = 252
    
        # ============================================================
        # 1. RETORNO LOGARÍTMICO
        # ============================================================
    
        df["Retorno_Log"] = np.log(df[i] / df[i].shift(1))
    
        # ============================================================
        # 2. VOLATILIDADE REALIZADA
        # ============================================================
    
        df["Vol_Realizada"] = df["Retorno_Log"].rolling(JANELA_VOL).std() * np.sqrt(DIAS_ANO) * 100
        df["Vol_20"] = df["Vol_Realizada"].quantile(0.20)
        df["Vol_80"] = df["Vol_Realizada"].quantile(0.80)
    
        # ============================================================
        # 3. RANK DA VOLATILIDADE
        # ============================================================
    
        vol_min = df["Vol_Realizada"].rolling(JANELA_IV).min()
        vol_max = df["Vol_Realizada"].rolling(JANELA_IV).max()
        denominador = vol_max - vol_min
    
        df["Vol_Rank"] = np.where(denominador != 0, ((df["Vol_Realizada"] - vol_min) / denominador) * 100, np.nan)
        df["Vol_Rank_20"] = df["Vol_Rank"].quantile(0.20)
        df["Vol_Rank_80"] = df["Vol_Rank"].quantile(0.80)
    
        # ============================================================
        # 4. PERCENTIL DA VOLATILIDADE
        # ============================================================
    
        df["Vol_Percentil"] = df["Vol_Realizada"].rolling(JANELA_IV).rank(pct=True) * 100
        df["Vol_Perc_20"] = df["Vol_Percentil"].quantile(0.20)
        df["Vol_Perc_80"] = df["Vol_Percentil"].quantile(0.80)
    
        # ============================================================
        # CLASSIFICAR VOLATILIDADE
        # ============================================================
    
        df["Vol_Hist"] = np.select([df["Vol_Realizada"] >= df["Vol_80"], df["Vol_Realizada"] <= df["Vol_20"]], ["Alta", "Baixa"], default="Neutra")
        df["Vol_Perc"] = np.select([df["Vol_Percentil"] >= df["Vol_Perc_80"], df["Vol_Percentil"] <= df["Vol_Perc_20"]], ["Alta", "Baixa"], default="Neutra")
        df["Vol_Rank"] = np.select([df["Vol_Rank"] >= df["Vol_Rank_80"], df["Vol_Rank"] <= df["Vol_Rank_20"]], ["Alta", "Baixa"], default="Neutra")
    
        # ============================================================
        # RESULTADO
        # ============================================================
    
        df = pd.DataFrame(df[['ticker', 'Vol_Rank', 'Vol_Perc', 'Vol_Hist']].iloc[-1]).T
        df_class_vol = pd.concat([df_class_vol, df], axis=0).reset_index(drop=True)
    
    # ============================================================
    # RENOMEAR ÍNDICE
    # ============================================================
    
    df_class_vol.columns = df_class_vol.columns.get_level_values(-1) if isinstance(df_class_vol.columns, pd.MultiIndex) else df_class_vol.columns
    df_class_vol = pd.DataFrame(df_class_vol.to_numpy(), columns=df_class_vol.columns)
    df_class_vol.index = range(len(df_class_vol))
    
    return df_class_vol
```

```python
def loog_classificacao_vol(tickers, max_workers=None):
    import pandas as pd
    import numpy as np
    from concurrent.futures import ThreadPoolExecutor

    # Coletar cotacoes
    df_lista_cotacoes = cotacoes_mt5_lista_2(tickers, timeframe=mt5.TIMEFRAME_D1, n_barras=600)

    JANELA_VOL = 21
    JANELA_IV = 252
    DIAS_ANO = 252

    def processar_ticker(i):
        try:
            df = pd.DataFrame(df_lista_cotacoes[i]).copy()
            df["ticker"] = i

            # Retorno logaritmico
            df["Retorno_Log"] = np.log(df[i] / df[i].shift(1))

            # Volatilidade realizada
            df["Vol_Realizada"] = df["Retorno_Log"].rolling(JANELA_VOL).std() * np.sqrt(DIAS_ANO) * 100
            df["Vol_20"] = df["Vol_Realizada"].quantile(0.20)
            df["Vol_80"] = df["Vol_Realizada"].quantile(0.80)

            # Rank da volatilidade
            vol_min = df["Vol_Realizada"].rolling(JANELA_IV).min()
            vol_max = df["Vol_Realizada"].rolling(JANELA_IV).max()
            denominador = vol_max - vol_min

            df["Vol_Rank"] = np.where(denominador != 0, ((df["Vol_Realizada"] - vol_min) / denominador) * 100, np.nan)
            df["Vol_Rank_20"] = df["Vol_Rank"].quantile(0.20)
            df["Vol_Rank_80"] = df["Vol_Rank"].quantile(0.80)

            # Percentil da volatilidade
            df["Vol_Percentil"] = df["Vol_Realizada"].rolling(JANELA_IV).rank(pct=True) * 100
            df["Vol_Perc_20"] = df["Vol_Percentil"].quantile(0.20)
            df["Vol_Perc_80"] = df["Vol_Percentil"].quantile(0.80)

            # Classificar volatilidade
            vol_hist = np.select([df["Vol_Realizada"] >= df["Vol_80"], df["Vol_Realizada"] <= df["Vol_20"]], ["Alta", "Baixa"], default="Neutra")
            vol_perc = np.select([df["Vol_Percentil"] >= df["Vol_Perc_80"], df["Vol_Percentil"] <= df["Vol_Perc_20"]], ["Alta", "Baixa"], default="Neutra")
            vol_rank = np.select([df["Vol_Rank"] >= df["Vol_Rank_80"], df["Vol_Rank"] <= df["Vol_Rank_20"]], ["Alta", "Baixa"], default="Neutra")

            # Resultado
            return pd.DataFrame({"ticker": [i], "Vol_Rank": [vol_rank[-1]], "Vol_Perc": [vol_perc[-1]], "Vol_Hist": [vol_hist[-1]]})

        except Exception as e:
            print(f"Erro ao processar {i}: {e}")
            return pd.DataFrame({"ticker": [i], "Vol_Rank": [np.nan], "Vol_Perc": [np.nan], "Vol_Hist": [np.nan]})

    # Paralelismo
    if max_workers is None:
        max_workers = min(32, len(df_lista_cotacoes))

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        resultados = list(executor.map(processar_ticker, df_lista_cotacoes.keys()))

    # Consolidar
    df_class_vol = pd.concat(resultados, ignore_index=True)

    return df_class_vol

df_class_vol = loog_classificacao_vol(tickers, max_workers=8)
df_class_vol
```

```python
def cotacoes_mt5_lista_2(tickers, timeframe=mt5.TIMEFRAME_D1, n_barras=600, tentativas=5):
    import pandas as pd
    import MetaTrader5 as mt5
    from datetime import datetime, timedelta
    import time

    if not mt5.initialize():
        raise RuntimeError(f"Erro ao conectar no MT5:\n{mt5.last_error()}")

    resultados = {}

    data_fim = datetime.now()
    data_inicio = data_fim - timedelta(days=int(n_barras * 3))

    for ticker in tickers:

        try:
            if not mt5.symbol_select(ticker, True):
                print(f"Ticker não encontrado: {ticker}")
                continue

            rates = None

            for tentativa in range(tentativas):

                rates = mt5.copy_rates_range(
                    ticker,
                    timeframe,
                    data_inicio,
                    data_fim
                )

                if rates is not None and len(rates) >= n_barras:
                    break

                time.sleep(0.5)

            if rates is None or len(rates) == 0:
                print(f"{ticker}: sem cotações")
                continue

            df = pd.DataFrame(rates)

            df["time"] = pd.to_datetime(df["time"], unit="s")

            #df = df.sort_values("time")
    
            #df = df.tail(n_barras)
            df = df.set_index('time')

            if len(df) < n_barras:
                print(f"{ticker}: {len(df)} barras")

            resultados[ticker] = df["close"].reset_index(drop=True)

        except Exception as e:
            print(f"Erro {ticker}: {e}")

    mt5.shutdown()

    if not resultados:
        return pd.DataFrame()

    df_cotacoes = pd.DataFrame(resultados)

    return df_cotacoes
```

```python
import MetaTrader5 as mt5
def cotacoes_mt5_OHLC(
    ticker,
    timeframe=mt5.TIMEFRAME_D1,
    n_barras=1000):


    import pandas as pd
    from IPython.display import clear_output

    """
    Coleta OHLCV do MetaTrader 5.

    Parâmetros
    ----------
    ticker : str
        Ex.: 'PETR4', 'WIN$', 'DOL$', 'EURUSD'

    timeframe :
        mt5.TIMEFRAME_M1
        mt5.TIMEFRAME_M5
        mt5.TIMEFRAME_M15
        mt5.TIMEFRAME_M30
        mt5.TIMEFRAME_H1
        mt5.TIMEFRAME_H4
        mt5.TIMEFRAME_D1
        mt5.TIMEFRAME_W1
        mt5.TIMEFRAME_MN1

    n_barras : int
        Quantidade de candles

    Retorna
    -------
    DataFrame
    """

    if not mt5.initialize():
        raise RuntimeError(
            f"Erro ao conectar no MT5:\n{mt5.last_error()}"
        )

    if not mt5.symbol_select(ticker, True):
        mt5.shutdown()
        raise ValueError(f"Ticker '{ticker}' não encontrado.")

    rates = mt5.copy_rates_from_pos(
        ticker,
        timeframe,
        0,
        n_barras
    )

    mt5.shutdown()

    if rates is None:
        raise ValueError("Nenhum dado retornado.")

    df = pd.DataFrame(rates)

    df["time"] = pd.to_datetime(
        df["time"],
        unit="s"
    )

    df.rename(
        columns={
            "time":"Data",
            "open":"Open",
            "high":"High",
            "low":"Low",
            "close":"Close",
            "tick_volume":"Volume",
            "real_volume":"RealVolume",
            "spread":"Spread"
        },
        inplace=True
    )

    df.set_index("Data", inplace=True)

    return df.iloc[:,:5]

import MetaTrader5 as mt5
cotacoes_mt5_OHLC('PETR4',    timeframe=mt5.TIMEFRAME_D1,    n_barras=1000)
```

```python
for acao in df_tickers_acoes[:2]:
    print(acao, f'0=== COLETAR COTAÇÕES DE AÇÃO e OPÇÃO de {acao} =====0')

    # Definir Nome Do Arquivo
    arquivo = f"CotacoesAcoesOpcoes/df_OpcoesAcoes_Cotacoes_{acao}.xlsx"

    # Lê a aba existente
    df_opcoes = pd.read_excel(arquivo, sheet_name="opcoes", index_col=0)
    # Coletar Tickers De Opções
    tickers_opcoes = df_opcoes['Ticker'].to_list()

    # Looping de Coleta de Cotações de Opções
    for tick in tickers_opcoes:
        #print(tick)

        try:
            cotacao_opcao = cotacoes_mt5_OHLC(tick, timeframe=mt5.TIMEFRAME_D1, n_barras=1)['Close']

            # Se retornou cotação
            if len(cotacao_opcao) > 0:
                valor_ultimo = cotacao_opcao.iloc[-1]
            else:
                valor_ultimo = 0

        except Exception:
            valor_ultimo = 0

        # Atualiza a coluna Último na linha do ticker
        df_opcoes.loc[df_opcoes['Ticker'] == tick, 'Último'] = valor_ultimo

    # Salvar arquivo mantendo a aba opcoes
    with pd.ExcelWriter(arquivo, engine="openpyxl", mode="a", if_sheet_exists="replace") as writer:
        df_opcoes.to_excel(writer, sheet_name="opcoes", index=True)
```
